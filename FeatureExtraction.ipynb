```python
def extract_episode(txt):    # ekhane anime gular episode sonkhake alada kora hoyecche.
    data = ""
    for i in range(len(txt)):
        if txt[i] == "(":
            for j in range(i+1, i+3): 
                data += txt[j]
            return data
```

```python
df["Episods"] = df['Title'].apply(extract_episode)  # prothome to function er madhome kontar kono episode choltese oita cound kora hoise 
df                                                    # and then tader ke notun ekta column er modhe rakha hoise jetar name "Episode"
```

```python
def extract_date(txt): 
    data = "" 
    for i in range(len(txt)): 
        if txt[i] == ")": 
            for j in range(i+1, i+20): 
                data += txt[j]
            return data
```

```python
df["Date"] = df["Title"].apply(extract_date)
```

```python
# Which is the longest running anime
```

```python
df["Start Date"] = pd.to_datetime(
    df["Date"].str.split(" - ").str[0],      # chatgpt theke newa hoyecche
    format="%b %Y",
    errors="coerce"
)

df["End Date"] = pd.to_datetime(
    df["Date"].str.split(" - ").str[1],
    format="%b %Y",
    errors="coerce"
)

df["Running Months"] = (
    (df["End Date"].dt.year - df["Start Date"].dt.year) * 12
    + df["End Date"].dt.month - df["Start Date"].dt.month
)

df.sort_values(by="Running Months", ascending=False).head(5)
```
